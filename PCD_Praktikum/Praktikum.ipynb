```python
import imageio.v3 as img
import numpy as np
import matplotlib.pyplot as plt

def Translasi(image, shiftX, shiftY):

    imgTranslasi = np.roll(image, shift=shiftY, axis=0) # Geser vertikal
    imgTranslasi = np.roll(imgTranslasi, shift=shiftX, axis=1) # Geser horizontal

    # Mengisi bagian yang kosong dengan warna hitam (0)
    if shiftY > 0:
        imgTranslasi[:shiftY, :] = 0 # Bagian atas jika geser ke bawah
    elif shiftY < 0:
        imgTranslasi[shiftY:, :] = 0 # Bagian bawah jika geser ke atas
    if shiftX > 0:
        imgTranslasi[:, :shiftX] = 0 # Bagian kiri jika geser ke kanan
    elif shiftX < 0:
        imgTranslasi[:, shiftX:] = 0 # Bagian kanan jika geser ke kiri

    return imgTranslasi


image = img.imread("WhatsApp Image 2026-06-13 at 11.34.44.jpeg")

imgResult = Translasi(image, shiftX=50, shiftY=-100)

plt.figure(figsize=(5,5))
plt.subplot(2,1,1)
plt.imshow(image)
plt.subplot(2,1,2)
plt.imshow(imgResult)
plt.show()
```

```python
import numpy as np
import imageio.v3 as img
import matplotlib.pyplot as plt

path = 'WhatsApp Image 2026-06-13 at 11.34.44.jpeg'
image = img.imread(path)

height, width = image.shape[:2]
horizontal = np.zeros_like(image)
vertical = np.zeros_like(image)
both = np.zeros_like(image)  # Mirroring Vertikal + Horizontal bersamaan

# Mirroring Horizontal
for y in range(height):
    for x in range(width):
        horizontal[y, x] = image[y, width - 1 - x]

# Mirroring Vertikal
for y in range(height):
    for x in range(width):
        vertical[y, x] = image[height - 1 - y, x]

# Mirroring Vertikal & Horizontal (Kombinasi)
for y in range(height):
    for x in range(width):
        both[y, x] = image[height - 1 - y, width - 1 - x]

# Menampilkan hasil
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.title("Gambar Asli")
plt.imshow(image)

plt.subplot(2, 2, 2)
plt.title("Mirroring Horizontal")
plt.imshow(horizontal)

plt.subplot(2, 2, 3)
plt.title("Mirroring Vertical")
plt.imshow(vertical)

plt.subplot(2, 2, 4)
plt.title("Mirroring Vertical & Horizontal")
plt.imshow(both)

plt.tight_layout()
plt.show()
```

```python
import imageio.v3 as img
import numpy as np
import matplotlib.pyplot as plt

def rotateImagePivotTopLeft(image, degree):
    radian_deg = np.radians(degree)
    cos_deg, sin_deg = np.cos(radian_deg), np.sin(radian_deg)

    height, width = image.shape[:2]

    # Menghitung dimensi matriks baru berdasarkan perputaran dari titik (0,0)
    max_dim = int(np.sqrt(height**2 + width**2))
    outputImage = np.zeros((max_dim, max_dim, 3), dtype=image.dtype)

    # Transformasi rotasi dengan titik pivot (0,0) di pojok kiri atas
    for y in range(height):
        for x in range(width):
            newX = int(cos_deg * x - sin_deg * y)
            newY = int(sin_deg * x + cos_deg * y)

            # Memastikan koordinat hasil rotasi berada di dalam batas matriks output
            if 0 <= newX < max_dim and 0 <= newY < max_dim:
                outputImage[newY, newX] = image[y, x]

    return outputImage

# Membaca gambar yang ada di sesi Colab
image = img.imread('WhatsApp Image 2026-06-13 at 11.34.44.jpeg')

# Memanggil fungsi rotasi
rotated_image = rotateImagePivotTopLeft(image, 45)

# Menampilkan hasil
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title("Gambar Asli")
plt.imshow(image)

plt.subplot(1, 2, 2)
plt.title("Rotasi (Pivot 0,0)")
plt.imshow(rotated_image)

plt.show()
```

```python
import numpy as np
import imageio.v3 as img
import matplotlib.pyplot as plt

# Fungsi Zoom Plus (Memperbesar)
def zoomPlus(image, factor):
    height, width = image.shape[:2]
    new_height = int(height / factor)
    new_width = int(width / factor)
    imgZoom = np.zeros((new_height, new_width, 3), dtype=image.dtype)

    for y in range(new_height):
        for x in range(new_width):
            ori_y = int(y * factor)
            ori_x = int(x * factor)

            ori_y = min(ori_y, height - 1)
            ori_x = min(ori_x, width - 1)

            imgZoom[y, x] = image[ori_y, ori_x]

    return imgZoom

# Fungsi Zoom Minus (Memperkecil Gambar)
def zoomMinus(image, factor):
    height, width = image.shape[:2]
    new_height = int(height * factor)  # Dimensi diperkecil
    new_width = int(width * factor)
    imgZoom = np.zeros((new_height, new_width, 3), dtype=image.dtype)

    for y in range(new_height):
        for x in range(new_width):
            ori_y = int(y / factor)
            ori_x = int(x / factor)

            ori_y = min(ori_y, height - 1)
            ori_x = min(ori_x, width - 1)

            imgZoom[y, x] = image[ori_y, ori_x]

    return imgZoom

# Membaca gambar yang ada di sesi Colab
image = img.imread('WhatsApp Image 2026-06-13 at 11.34.44.jpeg')

# Eksekusi Zooming
skala_plus = 2.0
skala_minus = 0.5  # Memperkecil gambar menjadi 50%

imgZoomedPlus = zoomPlus(image, skala_plus)
imgZoomedMinus = zoomMinus(image, skala_minus)

# Menampilkan hasil
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title("Gambar Asli")
plt.imshow(image)

plt.subplot(1, 3, 2)
plt.title("Zoom Plus (Cropped Zoom)")
plt.imshow(imgZoomedPlus)

plt.subplot(1, 3, 3)
plt.title("Zoom Minus (Di perbesar)")
plt.imshow(imgZoomedMinus)

plt.tight_layout()
plt.show()
```
